

### ✅ Easy Level
**Q1. What is bias in machine learning?**  
- **A:** Bias is the error due to simplifying assumptions in the model. High bias means the model underfits the data.

**Q2. What is variance in machine learning?**  
- **A:** Variance measures how much the model’s predictions change if the training data changes. High variance means the model overfits.

**Q3. What happens to bias and variance as model complexity increases?**  
- **A:** Bias decreases, variance increases.

---

### ⚖️ Moderate Level
**Q4. Explain the bias–variance tradeoff.**  
- **A:** As complexity increases, bias decreases but variance increases. The optimal model minimizes total error (bias² + variance + irreducible noise).

**Q5. Why does overfitting occur?**  
- **A:** Because a highly complex model fits noise in the training data, leading to high variance and poor generalization.

**Q6. What is underfitting?**  
- **A:** When a model is too simple, it cannot capture the underlying patterns → high bias, low variance.

---

### 🔹 Harder Level
**Q7. Is variance equal to test error?**  
- **A:** No. Test error = bias² + variance + irreducible noise. Variance is only one component.

**Q8. How do you reduce variance without increasing bias too much?**  
- **A:** Techniques like regularization (L1/L2), pruning decision trees, dropout in neural nets, or using ensemble methods (bagging, random forests).

**Q9. Give a real-world example of bias–variance tradeoff.**  
- **A:** Predicting house prices:  
  - A linear model (low complexity) → high bias, misses nonlinear trends.  
  - A very deep decision tree → high variance, fits noise.  
  - A random forest → balances bias and variance.

**Q10. How does cross-validation help with bias–variance tradeoff?**  
- **A:** It estimates test error more reliably, helping choose the right complexity that generalizes well.
